```python
import pandas as pd
import numpy as np
```

```python
def rle_encode(mask):
    pixels = mask.flatten(order='F')
    pixels = np.concatenate([[0], pixels, [0]])
    runs = np.where(pixels[1:] != pixels[:-1])[0]
    runs[1::2] -= runs[::2]
    return ' '.join(str(x) for x in runs)

rle_ones = rle_encode(np.ones((720, 1280), dtype=np.uint8))
rle_zeros = rle_encode(np.zeros((720, 1280), dtype=np.uint8))

def predict_dummy():
    test_ids = pd.read_csv('/kaggle/input/competitions/open-cv-py-torch-segmentation-project-round-3/test.csv')['ImageID'].to_numpy()
    image_id_col = []
    rle_col = []
    for image_id in test_ids:
        for class_idx in range(12):
            image_id_col.append(f'{image_id}_{class_idx}')
            rle_col.append(rle_ones if class_idx == 0 else rle_zeros)

    submission_df = pd.DataFrame({'ImageID': image_id_col, 'EncodedPixels': rle_col})
    submission_df.to_csv('/kaggle/working/submission.csv', index=False)
```

```python
predict_dummy()
```
