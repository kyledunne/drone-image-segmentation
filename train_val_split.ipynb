```python
import pandas as pd
import numpy as np
```

```python
train = pd.read_csv('data/train.csv')
cluster_labels = np.load('data_gen/cluster_labels.npy')
```

```python
train
```

```python
cluster_labels
```

```python
train['cluster'] = cluster_labels
train
```

```python
train.to_csv('data/train_with_clusters.csv', index=False)
```

```python
train_clusters = [1, 3, 4, 6, 7, 8]
val_clusters = [0, 2, 5, 9]
```

```python
train_split = train.query('cluster in @train_clusters')
val_split = train.query('cluster in @val_clusters')
```

```python
train_split
```

```python
val_split
```

```python
train_split.to_csv('data/train_split_with_cluster.csv', index=False)
val_split.to_csv('data/val_split_with_cluster.csv', index=False)
train_split['ImageID'].to_csv('data/train_split.csv', index=False)
val_split['ImageID'].to_csv('data/val_split.csv', index=False)
```
